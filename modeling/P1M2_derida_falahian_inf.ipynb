## **Model Inference**

```python
import joblib
import pandas as pd
```

```python
loaded_model = joblib.load(
    'best_model.pkl'
)
```

```python
data_inf = pd.DataFrame({
    'product': ['Mortgage'],
    'issue': ['Loan servicing'],
    'company': ['ABC Financial'],
    'state': ['CA'],
    'submitted_via': ['Web'],
    'company_response_to_consumer': ['Closed with explanation'],
    'timely_response': ['Yes'],
    'response_days':[5],
    'received_day':[10],
    'received_week':[2],
    'received_month':[11],
    'month_in_quarter':[4]
})
```

```python
pred = loaded_model.predict(data_inf)[0]

if pred == 1:
    print('Complaint diprediksi akan menjadi dispute')
else:
    print('Complaint diprediksi tidak akan menjadi dispute')
```
